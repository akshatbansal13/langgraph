```python
from langgraph.graph import StateGraph,START , END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
)
```

```python
class blogstate(TypedDict):
    topic : str
    outline: str
    content:str
```

```python
def outline(state:blogstate) ->blogstate:
    title = state['topic']
    prompt = f'Generate a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content
    state['outline'] = outline
    
    return state
```

```python
def blog(state:blogstate) ->blogstate:
    title = state['topic']
    outline = state['outline']

    prompt = f'Write a detailed blog on the title - {title} using the follwing outline \n {outline}'

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
graph = StateGraph(blogstate)

#add nodes
graph.add_node('outline',outline)
graph.add_node('content',blog)

#add edge
graph.add_edge(START , 'outline')
graph.add_edge('outline' , 'content')
graph.add_edge('content' , END)

workflow = graph.compile()
```

```python
intial_state = {'topic': 'Rise of AI in India'}

final_state = workflow.invoke(intial_state)

print(final_state)
```
